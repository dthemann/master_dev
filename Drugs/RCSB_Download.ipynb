```python
import urllib.request
import subprocess
from pathlib import Path

PDB_ID = "8CI0"
LIGAND_ID = "8EL"

# Download structure as mmCIF (RCSB no longer serves all entries as .pdb)
cif_file = f"{PDB_ID}.cif"
pdb_file = f"{PDB_ID}.pdb"

print(f"Downloading {PDB_ID} as mmCIF...")
urllib.request.urlretrieve(
    f"https://files.rcsb.org/download/{cif_file}",
    cif_file,
)
print(f"  ✓ {cif_file}")

# Convert CIF → PDB using OpenBabel
print(f"Converting {cif_file} → {pdb_file}...")
result = subprocess.run(
    ["obabel", cif_file, "-O", pdb_file],
    capture_output=True, text=True,
)
if Path(pdb_file).exists() and Path(pdb_file).stat().st_size > 0:
    print(f"  ✓ {pdb_file}")
else:
    print(f"  ✗ Conversion failed: {result.stderr.strip()}")

# Download ligand ideal coordinates
sdf_file = f"{LIGAND_ID}_ideal.sdf"
print(f"Downloading ligand {LIGAND_ID}...")
urllib.request.urlretrieve(
    f"https://files.rcsb.org/ligands/download/{sdf_file}",
    sdf_file,
)
print(f"  ✓ {sdf_file}")
```
